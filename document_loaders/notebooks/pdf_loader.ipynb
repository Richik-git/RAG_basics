```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("../knowledge-source/attention_is_all_you_need.pdf")
print(file_path.exists())
```

```python
# create a loader

pypdf_loader = PyPDFLoader(file_path=file_path.as_posix(),
                           mode='page')
```

```python
# load the documents

documents = pypdf_loader.load()
```

```python
print(type(documents))
print(len(documents))
```

```python
# fetch the first page text content

print(documents[0].page_content)
```

```python
# metadata

pp(documents[1].metadata)
```

```python
from langchain_community.document_loaders.parsers import TesseractBlobParser,RapidOCRBlobParser
```

```python
# create pypdf instance which can extract images

pypdf_image_loader = PyPDFLoader(file_path=file_path.as_posix(),
                                 mode='page',
                                 extract_images=True,
                                 images_parser=TesseractBlobParser(),
                                 images_inner_format='html-img')
```

```python
# load documents with images

doc_with_images = pypdf_image_loader.load()
```

```python
len(doc_with_images)
```

```python
page_with_img = doc_with_images[2]
print(page_with_img.page_content)
```

## PDF Miner

```python
from langchain_community.document_loaders import PDFMinerLoader
```

```python
# create the laoder

pdfminer_loader = PDFMinerLoader(file_path=file_path.as_posix(),
                                 mode='page',
                                 extract_images=True,
                                 images_parser=RapidOCRBlobParser(),
                                 images_inner_format='html-img')
```

```python
#load the doc

doc_with_images = pdfminer_loader.load()
```

```python
# image text to pdf

print(doc_with_images[2].page_content[-450:])
```

```python
print(doc_with_images[5].page_content)
```

```python
pp(doc_with_images[0].metadata)
```

```python
for doc in doc_with_images:
    if doc.metadata['page'] == 7:
        print(doc.page_content)
        break
```
