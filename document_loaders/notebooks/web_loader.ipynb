## Recursive loader

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
# create the loader

base_url = "https://docs.langchain.com/oss/python/integrations/document_loaders"

rec_loader = RecursiveUrlLoader(url=base_url,
                                max_depth=2)
```

```python
# load the docs

docs = rec_loader.load()
```

```python
len(docs)
```

```python
# metadata

for i in range(10):
    doc = docs[i]
    print(doc.metadata,end='\n\n')
```

```python
# lazy load

documents_lazy_load = rec_loader.lazy_load()
```

```python
counter = 0

for document in documents_lazy_load:
    # stop condition
    if counter == 20:
        break
    
    counter += 1
    
    print(document.page_content[0:300])
    
    pp(document.metadata)
```
